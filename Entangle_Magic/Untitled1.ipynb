```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from scipy.linalg import expm
from itertools import product
```

```python
d_logf = 0.51

def d_overlap(t,overlap):
    return -d_logf*overlap

t_list = np.linspace(0,10.0,101)
soln = solve_ivp(d_overlap, t_span=[t_list[0],t_list[-1]], y0=[0.001], method='RK45', t_eval=t_list)

plt.plot(soln.t,soln.y[0])
```
