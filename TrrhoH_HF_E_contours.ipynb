```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from scipy.linalg import expm
from itertools import product
```

```python
'''
This is the energy contour based on <Phi|H|Phi> in Wang Cassing
'''

eps = 1
V = 4
# N = 3

delta = 0.005
th = np.arange(0.0,np.pi,delta)
ph = np.arange(0.0,2*np.pi,2*delta)
Th,Ph = np.meshgrid(th,ph)
# E is the energy for N=3
E = -3/4*( 2*eps*np.cos(Th) + V*np.cos(2*Ph)*np.sin(Th)**2 )

fig, ax = plt.subplots()
CS = ax.contour(Th, Ph, E)
ax.clabel(CS, fontsize=10)
ax.set_title(r'$E_{HF}$')
ax.set_xlabel(r'$\theta$')
ax.set_ylabel(r'$\phi$')
ax.plot(np.arccos(eps/V),0,marker='X',color='r')
ax.plot(-np.arccos(eps/V)+np.pi*2,0,marker='X',color='green') # outside domain of 
ax.plot(np.arccos(-eps/V),np.pi/2,marker='X',color='b')
ax.plot(np.arccos(eps/V),np.pi,marker='X',color='r')
```
