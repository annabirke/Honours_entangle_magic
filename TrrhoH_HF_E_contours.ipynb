```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from scipy.linalg import expm
from itertools import product
```

```python
'''
This is the energy contour based on <Phi|H|Phi>, gives the same result as Tr(H rho)
'''

eps = 1
V = 4
N = 3

delta = 0.005
th = np.arange(0.0,np.pi,delta)
ph = np.arange(0.0,2*np.pi,2*delta)
Th,Ph = np.meshgrid(th,ph)
# E is the energy for N=3
E = -N/2*( eps*np.cos(Th) + V*(N-1)/2 *np.cos(2*Ph)*np.sin(Th)**2 )

fig, ax = plt.subplots()
CS = ax.contour(Th, Ph, E)
ax.clabel(CS, fontsize=10)
ax.set_title(r'$E_{HF}$')
ax.set_xlabel(r'$\theta$')
ax.set_ylabel(r'$\phi$')
ax.plot(np.arccos(eps/(V*N-V)),0,marker='X',color='r')
# ax.plot(-np.arccos(eps/(V*N-V))+2*np.pi,0,marker='X',color='green') # outside domain of theta
# ax.plot(np.arccos(-eps/(V*N-V)),np.pi/2,marker='X',color='b')
ax.plot(np.arccos(eps/(V*N-V)),np.pi,marker='X',color='r')
# ax.set_ylim(-0.3,3.4)
# plt.savefig(f'C:/Users/annas/Documents/2026/Honours/Entangle_Magic/lipkin_EHF_contours.png',bbox_inches='tight',dpi=300)
plt.show()
```
