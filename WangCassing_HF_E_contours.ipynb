```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from scipy.linalg import expm
from itertools import product
```

```python
'''
This is the energy contour based on <Phi|H|Phi> in Wang Cassing
'''

eps = 1
V = 4
N = 3

delta = 0.005
th = np.arange(0.0,np.pi,delta)
ph = np.arange(0.0,2*np.pi,2*delta)
X,Y = np.meshgrid(th,ph)
# Z is the energy for N=3
Z = (N*eps/2)*np.cos(X) - (3*V/2)*(np.sin(X)**2)*np.cos(2*Y)

fig, ax = plt.subplots()
CS = ax.contour(X, Y, Z)
ax.clabel(CS, fontsize=10)
ax.set_title(r'$E_{HF}$')
ax.set_xlabel(r'$\theta$')
ax.set_ylabel(r'$\phi$')
```
